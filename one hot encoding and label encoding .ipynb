```python
from sklearn.preprocessing import PowerTransformer
```

```python
# A right skewed feature
x = np.array([ [-1],[9],[0],[5],[10],[50]   ])
```

```python
# create the transformer
pt = PowerTransformer(method="yeo-johnson")
```

```python
# fit and transform
x_transformed = pt.fit_transform(x)
```

```python
# A right skewed feature
y = np.array([ [1],[9],[6],[5],[10],[50] ])
```

```python
#create the transformer
pt = PowerTransformer(method="box-cox")
```

```python
#fit and transform
y_transformed = pt.fit_transform(y)
```

```python
import pandas as pd
import numpy as np

data = {
    'Employee_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Department': ['IT', 'HR', 'Finance', None, 'IT', 'Sales', 'HR', None, 'IT', 'Finance'],
    'City': ['Bhopal', 'Indore', np.nan, 'Delhi', 'Mumbai', 'Bhopal', 'Pune', 'Indore', np.nan, 'Delhi'],
    'Education': ['B.Tech', 'MBA', 'B.Com', 'B.Tech', None, 'MBA', 'B.Sc', 'B.Tech', 'MBA', None],
    'Age': [22, 25, np.nan, 28, 30,  np.nan, 26, 32, 24, 29],
    'Salary': [25000, 35000, 40000, np.nan, 50000, 45000, np.nan, 60000, 30000, 55000],
    'Experience': [0, 2, 3, 5, np.nan, 4, 1, 7, 2, np.nan],
    'Year': [2024, 2023, 2022, 2021, 2024, np.nan, 2023, 2020, 2022, 2021]
}

df = pd.DataFrame(data)
print(df)
```

```python
from sklearn.impute import SimpleImputer
```

```python
imputer = SimpleImputer(strategy= 'mean')
imputer.fit_transform(df[['Salary']])
```

```python
imputer1 = SimpleImputer(strategy='mean')
df['new_salary']= imputer1.fit_transform(df[['Salary']])
```

```python
imputer2 = SimpleImputer(strategy= 'median')
df['new_experience']=imputer2.fit_transform(df[['Experience']])
df
```

```python
imputer3 = SimpleImputer(strategy='most_frequent')
df['City'] = imputer3.fit_transform(df[['City']]).flatten()
```
